# Imports

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

# Datas

```python
players = pd.read_csv('C:/Users/Lenovo/OneDrive/Desktop/Desk/Quera/First-Project/players.csv')
team_rosters = pd.read_csv('C:/Users/Lenovo/OneDrive/Desktop/Desk/Quera/First-Project/team_rosters.csv')
mvp_votes = pd.read_csv('C:/Users/Lenovo/OneDrive/Desktop/Desk/Quera/First-Project/mvp_votes.csv')
player_season = pd.read_csv('C:/Users/Lenovo/OneDrive/Desktop/Desk/Quera/First-Project/player_seasons.csv')
champions = pd.read_csv('C:/Users/Lenovo/OneDrive/Desktop/Desk/Quera/First-Project/champions.csv')
```

# team

```python
df_player_season = player_season.copy()
teams = df_player_season['team_id'].unique().dropna()
teams = np.sort(teams)
lst= []
for i in range(len(teams)):
    lst.append(i+1)
df_team = pd.DataFrame({
    'team_id': lst,
    'team_label': teams
})
df_team
```

# Champion

```python
champions.head(10)
```

```python
df_champ = champions.copy()
seasons = df_champ['season'].unique()

for i, seas in enumerate(seasons, start=1):
    df_champ['season'] = df_champ['season'].replace(seas, i)
df_champ = df_champ.rename(columns= {'season': 'season_id'})
df_champ['team_id'] = df_champ['team_id'].map(df_team.set_index('team_label')['team_id'])

df_champ = df_champ[['season_id','team_id']]
df_champ
```

# Season

```python
df_season = champions.copy()
df_season['season_id'] = df_champ['season_id']
df_season = df_season[['season_id','season']]
df_season
```

# Positions

```python
team_rosters.head(10)
```

```python

positions = team_rosters['position'].unique()
positions = np.sort(positions)
lst_pose=[1,2,3,4,5]
df_positions = pd.DataFrame({
        'position_id':lst_pose,
    'position' : positions
})
df_positions
```

# Experience

```python
df_experience = team_rosters.copy()
df_experience = df_experience.rename(columns= {'position': 'position_id'})
df_experience = df_experience.rename(columns= {'season': 'season_id'})
df_experience['team_id'] = df_experience['team_id'].map(df_team.set_index('team_label')['team_id'])
df_experience['season_id'] = df_experience['season_id'].map(df_season.set_index('season')['season_id'])
df_experience['position_id'] = df_experience['position_id'].map(df_positions.set_index('position')['position_id'])
lst_exp= []
for i in range(len(df_experience['season_id'])):
    lst_exp.append(i+1)
df_experienced = pd.DataFrame({'experience_id' : lst_exp})
df_experience['experience_id'] = df_experienced['experience_id']
df_experience= df_experience[['experience_id','season_id','team_id','player_id','position_id','experience_years']]
```

# Players

```python
players.head(10)
```

```python
df_player = players.copy()
df_player = df_player[['player_id','name','birth_date','height_cm','weight_kg','birth_country']]
```

# MVP_Votes

```python
mvp_votes.head(10)
```

```python
df_mvp = mvp_votes.copy()
df_mvp = df_mvp.rename(columns= {'season': 'season_id'})
df_mvp['season_id'] = df_mvp['season_id'].map(df_season.set_index('season')['season_id'])
lst_mvp =[]
for i in range(len(df_mvp['season_id'])):
    lst_mvp.append(i +1)
df_mv = pd.DataFrame({'mvp_id': lst_mvp})
df_mvp['mvp_id'] =df_mv['mvp_id']
df_mvp = df_mvp[['mvp_id','season_id','player_id','rank','first_place_votes','points_won','points_max','vote_share']]
df_mvp
```

# Player 0f seasons

```python
player_season.head(10)
```

```python
df_player_season = df_player_season.rename(columns= {'position': 'position_id'})
df_player_season = df_player_season.rename(columns= {'season': 'season_id'})
df_player_season['team_id'] = df_player_season['team_id'].map(df_team.set_index('team_label')['team_id'])
df_player_season['season_id'] = df_player_season['season_id'].map(df_season.set_index('season')['season_id'])
df_player_season['position_id'] = df_player_season['position_id'].map(df_positions.set_index('position')['position_id'])
df_player_season['is_total'] = df_player_season['is_total'].map({False : 0, True : 1})
lst_play = []
for i in range(len(df_player_season['season_id'])):
    lst_play.append(i + 1)
df_player_season.insert(0, 'player_season_id', lst_play)
df_player_season = df_player_season.drop(
    ['source_url', 'fetched_at', 'advanced_source_url', 'advanced_fetched_at'],
    axis=1
)

df_player_season
```

# New Datas
